______________
### Escalamiendo MinMax

```python
dfscaled = df.copy()
for col in dfscaled.columns: 
    minn = dfscaled[col].min()
    dfscaled[col] = (dfscaled[col]-minn)
    maxx = dfscaled[col].max()
    dfscaled[col]/=maxx
```

```python
dfscaled.describe()
```

___________
### Escalamiento 

```python
dfscaled2 = df.copy()
for col in dfscaled2.columns: 
    media = dfscaled2[col].mean()
    dfscaled2[col] = (dfscaled2[col]-media)
    std = dfscaled2[col].std()
    dfscaled2[col]/=std
```
