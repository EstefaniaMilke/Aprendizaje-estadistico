```python
import pandas as pd 
import numpy as np 
from sklearn.linear_model import LinearRegression
from scipy import stats
import statsmodels.api as sm
```

**Carga de datos**

```python
df0 = pd.read_excel(r"C:\Users\valer\aguasfeo.xlsx")
```

```python
df0.head()
```

**Modelo con codificación 0-1**

```python
dff = df0.copy()
df = pd.DataFrame({
    "Tamaño":      (dff["Tamaño"] == "grande").astype(int),
    "Hielo":    (dff["Hielo"] == "con").astype(int),
    "Kahlúa":   (dff["Sabor"] == "kahlúa").astype(int),
    "Horchata": (dff["Sabor"] == "horchata").astype(int)
})

df.head(10)
```

```python
x1 = df[["Tamaño","Hielo","Kahlúa","Horchata"]]
x = sm.add_constant(x1)
y = df0[["Score"]].values.reshape(-1,1)
```

```python
xd = sm.add_constant(x)
model = sm.OLS(y,xd)
results = model.fit()
results.summary()
```

______________________

**Modelo con codificación -1–1**

```python
df2 = pd.DataFrame({
    "Tamaño": df["Tamaño"].replace({1: -1, 0: 1}),
    "Hielo": df["Hielo"].replace({0: -1}),
    "Kahlúa": (dff["Sabor"] == "kahlúa").astype(int),
    "Horchata": (dff["Sabor"] == "horchata").astype(int)
})

df2.head(10)
```

```python
x2 = df[["Tamaño","Hielo","Kahlúa","Horchata"]]
xx = sm.add_constant(x2)
yy = df0[["Score"]].values.reshape(-1,1)
```

```python
xd2 = sm.add_constant(xx)
model2 = sm.OLS(yy,xd2)
results2 = model2.fit()
results2.summary()
```
