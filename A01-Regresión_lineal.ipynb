### Archivo Salarios Mínimos

**Regresión polinomial de grado 1**

*Matrices X y Y*

```python
x1 = salmin[["Periodo"]]
x1.values
```

```python
y1 = salmin[["Salario"]]
y1.values
```

**Construcción del modelo con sklearn**

```python
lr1 = LinearRegression()
lr1.fit(x1,y1)
```

```python
lr1.intercept_
```

```python
lr1.coef_
```

*Regresión Lineal*

y = -5022.60217164 + 2.52832885x
________________________________

**Cálculo de R^2**

```python
r2_1 = lr1.score(x1,y1)
print("El estadístico R2 es:",r2_1)
```

El modelo explica aproximadamente el 98.4 % de la variabilidad. Por lo tanto, el modelo ajusta adecuadamente los datos analizados.

*Regresión Lineal*

**Gráfico comparativo**

*Predicciones*

```python
ypred1 = lr1.predict(x1)
ypred2 = lr2.predict(x2)
ypred3 = lr3.predict(x3)
ypred4 = lr4.predict(x4)
```

```python
plt.figure()
plt.xlabel("Periodo"); plt.ylabel("Salario");plt.grid()
plt.title("Gráfico datos originales vs propuestas")
plt.scatter(salmin["Periodo"], salmin["Salario"], label="Datos originales")
# Líneas de los modelos
plt.plot(salmin["Periodo"], ypred1, label="Grado 1", color="purple")
plt.plot(salmin["Periodo"], ypred2, label="Grado 3", color="orange")
plt.plot(salmin["Periodo"], ypred3, label="Grado 5", color="green")
plt.plot(salmin["Periodo"], ypred4, label="Grado 7", color="red")

plt.legend(loc="best")
plt.show()
```

*Interpretación*

Los modelos polinomiales presentan valores altos de R^2, lo que indica un buen ajuste a los datos. No obstante, al aumentar el grado del polinomio, la mejora en el ajuste es marginal, por lo que un modelo de menor grado resulta suficiente para describir la tendencia general del salario mínimo, esto debido a que desde un inicio los datos ya presentan una tendencia casi linear.
