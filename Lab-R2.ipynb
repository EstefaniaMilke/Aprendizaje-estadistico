## Regresión polinomial

Observa la siguiente gráfica.

```python
import matplotlib.pyplot as plt; import numpy as np
%matplotlib inline
np.random.seed(0)
x = np.linspace(-3, 3, 100); y = 1 + 4*x**2 + np.random.normal(0, 1, 100)
plt.figure(); plt.xlabel("x"); plt.ylabel("y")
plt.scatter(x, y); plt.grid()
plt.show()
```

Utiliza `sklearn` para realizar una regresión lineal simple ($\hat{y} = \hat{\beta_0} + \hat{\beta_1}x $)

```python
from sklearn.linear_model import LinearRegression
```

```python
x = x.reshape(-1,1)
```

```python
y = y.reshape(-1,1)
```

```python
lr = LinearRegression()
lr.fit(x,y)
```

```python
lr.intercept_
```

```python
lr.coef_
```

Regresión

y = 13.30223226 - 0.04955448x

____________

Crea una gráfica donde se muestren las observaciones como una nube de puntos y la línea que ajustaste con regresión. Calcula el métrico $R^2$.

```python
ypred = lr.predict(x)
```

```python
plt.figure()
plt.xlabel("x"); plt.ylabel("y");plt.grid()
plt.title("Regresión Lineal")
plt.scatter(x, y, label="Datos originales",color="gray")
# Línea del modelo
plt.plot(x, ypred, label="Modelo", color="black")
plt.legend(loc="best")
plt.show()
```

```python
r2 = lr.score(x,y)
print("El estadístico R2 es:",r2)
```

____________

Podemos ver que el ajuste no es correcto, pues parece que tiene un comportamiento cuadrático y estamos intentando ajustar una línea recta.

En lugar de hacer una regresión lineal simple, quizás podamos extender el modelo para que tenga varios factores dentro:

$$ Y = \beta_0 + \beta_1 X + \beta_2 X^2 + ... + \beta_p X^p + \epsilon$$

Ajusta un modelo cuadrático con `sklearn` ($\hat{y} = \hat{\beta_0} + \hat{\beta_1}x + \hat{\beta_1}x^2 $) y calcula su $R^2$.

```python
x_2 = x**2
```

```python
x2 = np.hstack((x,x_2))
```

```python
lr2 = LinearRegression()
lr2.fit(x2,y)
```

```python
lr2.intercept_
```

```python
lr2.coef_
```

Regresión Lineal

y = 0.61942096 - 0.04955448x + 4.14388884x^2

Crea una gráfica con las observaciones y la nueva curva que ajustaste.

```python
ypred2 = lr2.predict(x2)
```

```python
plt.figure()
plt.xlabel("x"); plt.ylabel("y");plt.grid()
plt.title("Regresión Cuadrática")
plt.scatter(x, y, label="Datos originales",color="gray")
# Línea del modelo
plt.plot(x, ypred2, label="Modelo", color="black")
plt.legend(loc="best")
plt.show()
```

```python
r2 = lr2.score(x2,y)
print("El estadístico R2 es:",r2)
```

**¿Cuál modelo consideras que es mejor? ¿Por qué?**

Debido a que el modelo se construyó en base a la función y = 1 + 4x^2 + r (una función cuadrática), el mejor modelo para este caso es la regresión cuadrática debido a lo siguiente:
1. La gráfica muestra una forma muy samilar a los datos originales, lo que indica que representa adecuadamente la relación entre las variables
2. Obtiene un valor de R^2 = 0.9934, el cual es muy alto 

Como podemos observar, las gráficas y valores de R^2 no cambian significativamente en grados posteriores; por lo tanto, este modelo es el mejor debido a que explica bien la variabilidad de los datos, sin agregar complicaciones innecesarias. De esta forma, el modelo puede presentar un mejor comportamiento extrapolar, ya que los modelos de grados altos tienden a comportarse inestablemente fuera del dominio. 

Recuerda que una de las metas de ajustar modelos es para poder interpolar y extrapolar. Los datos que tenemos mantienen $x \in [-3, 3]$. ¿Qué pasaría si quisiéramos saber el valor de $y$ cuando $x=5$?

Calcula la predicción $\hat{y}$ para todos los modelos creados (grados 2, 3, 5, 7, 9, y 11).

```python
x_new = np.array([[5]])
```

```python
y2_5  = lr2.predict(np.hstack((x_new, x_new**2)))
y3_5  = lr3.predict(np.hstack((x_new, x_new**2, x_new**3)))
y5_5  = lr5.predict(np.hstack((x_new, x_new**2, x_new**3, x_new**4, x_new**5)))
y7_5  = lr7.predict(np.hstack((x_new, x_new**2, x_new**3, x_new**4, x_new**5, x_new**6, x_new**7)))
y9_5  = lr9.predict(np.hstack((x_new, x_new**2, x_new**3, x_new**4, x_new**5, x_new**6, x_new**7, x_new**8, x_new**9)))
y11_5 = lr11.predict(np.hstack((x_new, x_new**2, x_new**3, x_new**4, x_new**5, x_new**6, x_new**7, x_new**8, x_new**9, x_new**10, x_new**11)))
```

```python
print("Grado 2:", y2_5)
print("Grado 3:", y3_5)
print("Grado 5:", y5_5)
print("Grado 7:", y7_5)
print("Grado 9:", y9_5)
print("Grado 11:", y11_5)
```

¿Cuál modelo crees que sea mejor?

Si calculamos "y" con la función original, tenemos y = 1 + 4(5)^2, que es igual a 101. Si observamos las predicciones, los valores más cercanos a 101 son los de grado 2 y 3. En contraste, las regresiones de grado alto se vuelven inestables y producen resultados incongruentes, sobre todo la del grado 11. Por esta razón, el modelo cuadrático se considera el más adecuado, ya que logra un buen ajuste dentro del rango observado y mantiene un comportamiento estable al extrapolar. 
Haciendo este ejercicio, podemos concluir que un R^2 alto no garantiza predicciones adecuadas fuera del dominio original. En cierto sentido, la extrapolación revela la verdadera calidad del modelo.
